```python
import os
import shutil
import zipfile

os.makedirs("/kaggle/working/checkpoints", exist_ok=True)
os.makedirs("/kaggle/working/outputs", exist_ok=True)

print("Scanning for model...")
for root, dirs, files in os.walk("/kaggle/input"):
    # Physically block os.walk from entering the massive image folders
    dirs[:] = [d for d in dirs if "plantvillage" not in d.lower() and "abdallah" not in d.lower() and "code" not in d.lower()]
    
    for f in files:
        if f == "best_model.pth":
            src = os.path.join(root, f)
            print(f"Success! Found raw model at: {src}")
            shutil.copy2(src, "/kaggle/working/checkpoints/best_model.pth")
            
        elif "output" in f.lower() and f.endswith(".zip"):
            src = os.path.join(root, f)
            print(f"Found zip archive at: {src}, extracting...")
            # Copy to working dir first to bypass Kaggle's read-only unzip error
            shutil.copy2(src, "/kaggle/working/temp.zip")
            with zipfile.ZipFile("/kaggle/working/temp.zip", 'r') as zip_ref:
                zip_ref.extractall("/kaggle/working/")

print("\nVerification: Contents of /kaggle/working/checkpoints/")
os.system("ls -lh /kaggle/working/checkpoints/")
```
